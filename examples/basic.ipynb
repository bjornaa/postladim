# Postladim example

```python
# Imports
import matplotlib.pyplot as plt
import numpy as np

from postladim import ParticleFile
```

First open a LADiM output file, we will use the output from the ``line`` example in LADiM.
Modify the path to fit your LADiM installation.

```python
ladim_file = "/home/bjorn/ladim2/examples/line/out.nc"
```

```python
# Open the particle file
pf = ParticleFile(ladim_file)
```

Check the size of the data set

```python

print("Number of time steps:         ", pf.num_times)
print("Number of particles:          ", pf.num_particles)
print("Number of particle instances: ", pf.count.sum())
```

```python
# Distribution data at first time step
print("First time step:\n", pf.time[0])
```

```python
# Use ftime to more concise output
print("First time step: ", pf.ftime(0))
print("Last time step : ", pf.ftime(-1))
```

```python
step = pf.time[1] - pf.time[0]
# Convert from nanoseconds to hours
hour = np.timedelta64(1, "h")  # Duration of 1 hour
print(f"Time step size: {float(step / hour)} hours")
```

Get and plot initial position (blue) and final (red). For nicer plot with geographical coordinates, coast line and bottom contours look at the LADiM line example.

```python
# Get (and plot) the initial and final positions
X0, Y0 = pf.position(0)
plt.plot(X0, Y0, ".", color="blue");
X1, Y1 = pf.position(-1)
plt.plot(X1, Y1, ".", color="red");
```

Extract and plot every 10th trajectory.

```python
for p in np.arange(0, 1000, 10):
    X, Y = pf.trajectory(pid=p)
    plt.plot(X, Y)
```
